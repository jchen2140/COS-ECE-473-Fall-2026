## Swap TIGER for LUSD

We approve the Sepolia SwapRouter02 to spend up to 10 of each token. Each approval waits for a successful receipt before the next transaction is submitted. Gas limits are estimated for each call with a 20% buffer; fee caps use the current base fee and the provider’s suggested priority fee.

```python
def build_lab_transaction(function, wallet_address):
    # Clear a legacy strategy left over from earlier notebook runs.
    w3.eth.set_gas_price_strategy(None)
    # Estimate this call's gas, then allow a 20% buffer.
    gas_estimate = function.estimate_gas({'from': wallet_address})
    priority_fee = w3.eth.max_priority_fee
    base_fee = w3.eth.get_block('latest')['baseFeePerGas']
    return function.build_transaction({
        'from': wallet_address,
        'chainId': w3.eth.chain_id,
        'nonce': w3.eth.get_transaction_count(wallet_address, 'pending'),
        'gas': (gas_estimate * 120 + 99) // 100,
        'maxPriorityFeePerGas': priority_fee,
        'maxFeePerGas': 2 * base_fee + priority_fee,
    })


def approve(token, abi_token, spender_address, wallet_address, private_key):
    max_amount = w3.to_wei(10, 'ether') # Allow up to 10 tokens for the lab
    token_contract = w3.eth.contract(token, abi=abi_token)
    tx = build_lab_transaction(
        token_contract.functions.approve(spender_address, max_amount), wallet_address
    )
    signed_tx = w3.eth.account.sign_transaction(tx, private_key)
    tx_hash = w3.eth.send_raw_transaction(signed_tx.raw_transaction)
    receipt = w3.eth.wait_for_transaction_receipt(tx_hash)
    assert receipt.status == 1, "Approval reverted"
    return w3.to_hex(tx_hash)
```

## Atomic swap transaction

- Use Uniswap’s multicall to invoke two swaps atomically: either both succeed or both revert. The second swap spends exactly 1 LUSD from your wallet, rather than using the first swap’s output amount.
- SwapRouter02 has no deadline field in its swap tuple; the deadline belongs to the multicall wrapper.
- These testnet examples use `amountOutMinimum = 0` to focus on transaction mechanics; this accepts any output amount.

```python
# Use the SwapRouter02 instance initialized above.
params1 = [{
    'tokenIn': tiger_token_address,
    'tokenOut': lusd_token_address,
    'fee': 500,
    'recipient': public_address,
    'amountIn': w3.to_wei(1, 'ether'), #Exchange 1 token
    'amountOutMinimum': 0,
    'sqrtPriceLimitX96': 0,
    }
    ]

params2 = [{
    'tokenIn': lusd_token_address,
    'tokenOut': tiger_token_address,
    'fee': 500,
    'recipient': public_address,
    'amountIn': w3.to_wei(1, 'ether'), #Exchange 1 token
    'amountOutMinimum': 0,
    'sqrtPriceLimitX96': 0,
    }
        ]


tx1 = router_contract.encode_abi('exactInputSingle',args=params1)
tx2 = router_contract.encode_abi('exactInputSingle',args=params2)
data = [tx1,tx2]

atomic_swap = router_contract.get_function_by_signature("multicall(uint256,bytes[])")(
    int(time.time()) + 1200, data
)
txn = build_lab_transaction(atomic_swap, public_address)
signed_txn = w3.eth.account.sign_transaction(txn, private_key)
txn_hash = w3.eth.send_raw_transaction(signed_txn.raw_transaction)
tx_hash_hex = w3.to_hex(txn_hash)
print("Hash of atomic swap transaction:", tx_hash_hex)
print(f"Sepolia Etherscan: https://sepolia.etherscan.io/tx/{tx_hash_hex}")
receipt = w3.eth.wait_for_transaction_receipt(txn_hash)
assert receipt.status == 1, "Swap reverted"
```
